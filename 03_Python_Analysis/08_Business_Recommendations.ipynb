# Notebook 8 — Business Recommendations
**Purpose:** Synthesize findings from Notebooks 1-7 into a ranked set of
business recommendations with quantified, clearly-labeled *simulated* impact
estimates, feeding directly into the Business Consulting Report and CEO deck.

**Libraries used:** `pandas`, `matplotlib`


```python
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = "../01_Data"
sales = pd.read_csv(f"{DATA_DIR}/sales_clean.csv", parse_dates=["Transaction_Date"])
inventory = pd.read_csv(f"{DATA_DIR}/inventory.csv")

monthly = sales.set_index('Transaction_Date').resample('MS').agg(
    Revenue=('Sales_Amount','sum'), Profit=('Profit','sum')
)
monthly['Margin_Pct'] = 100*monthly['Profit']/monthly['Revenue']

first_q_margin = monthly['Margin_Pct'].iloc[:3].mean()
latest_q_margin = monthly['Margin_Pct'].iloc[-3:].mean()
annualized_revenue = sales['Sales_Amount'].sum() / 2   # 2 years of data

margin_gap = first_q_margin - latest_q_margin
recovery_opportunity = annualized_revenue * (margin_gap/100)

print(f"Margin, first quarter observed : {first_q_margin:.1f}%")
print(f"Margin, latest quarter observed: {latest_q_margin:.1f}%")
print(f"Margin gap                     : {margin_gap:.1f} points")
print(f"Approx. annualized revenue base: NGN {annualized_revenue:,.0f}")
print(f"Estimated annual profitability recovery opportunity if margin gap is closed:")
print(f"   NGN {recovery_opportunity:,.0f}  (SIMULATED — directional sizing only)")
```

```python
slow_moving = inventory[inventory['Stock_Level'] > inventory['Reorder_Level']*2]
overstock_value = slow_moving['Inventory_Cost'].sum()
target_reduction_pct = 0.18
working_capital_release = overstock_value * target_reduction_pct

print(f"Inventory value tied up in slow-moving stock: NGN {overstock_value:,.0f}")
print(f"Target reduction (18%): NGN {working_capital_release:,.0f} working-capital release (SIMULATED)")

stockout_risk = inventory[inventory['Stock_Level'] <= inventory['Reorder_Level']]
print(f"\nProduct-store combinations at/below reorder level: {len(stockout_risk)} of {len(inventory)} "
      f"({100*len(stockout_risk)/len(inventory):.0f}%)")
```
