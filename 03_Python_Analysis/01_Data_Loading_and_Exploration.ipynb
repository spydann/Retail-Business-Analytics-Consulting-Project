# Notebook 1 — Data Loading & Exploration
**Engagement:** Retail Profitability & Performance Optimization — ApexMart Retail Ltd.
**Purpose:** Load all eight raw datasets, confirm structure, and perform an initial
structural inventory before any cleaning or analysis takes place.

**Libraries used:** `pandas`, `numpy`


```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', 20)
pd.set_option('display.width', 140)

DATA_DIR = "../01_Data"

sales      = pd.read_csv(f"{DATA_DIR}/sales_transactions.csv")
customers  = pd.read_csv(f"{DATA_DIR}/customers.csv")
products   = pd.read_csv(f"{DATA_DIR}/products.csv")
stores     = pd.read_csv(f"{DATA_DIR}/stores.csv")
inventory  = pd.read_csv(f"{DATA_DIR}/inventory.csv")
campaigns  = pd.read_csv(f"{DATA_DIR}/marketing_campaigns.csv")
returns    = pd.read_csv(f"{DATA_DIR}/returns.csv")
suppliers  = pd.read_csv(f"{DATA_DIR}/suppliers.csv")

print("Datasets loaded successfully.")
```

## 1.1 Dataset Row & Column Inventory

```python
datasets = {
    "sales_transactions": sales, "customers": customers, "products": products,
    "stores": stores, "inventory": inventory, "marketing_campaigns": campaigns,
    "returns": returns, "suppliers": suppliers
}

inventory_summary = pd.DataFrame([
    {"Dataset": name, "Rows": df.shape[0], "Columns": df.shape[1]}
    for name, df in datasets.items()
])
print(inventory_summary.to_string(index=False))
```

## 1.2 Schema Preview — Sales Transactions (core fact table)

```python
print(sales.dtypes)
print()
sales.head(8)
```

## 1.3 Date Range Covered

```python
sales['Transaction_Date'] = pd.to_datetime(sales['Transaction_Date'])
print("Earliest transaction:", sales['Transaction_Date'].min().date())
print("Latest transaction:  ", sales['Transaction_Date'].max().date())
print("Total transactions:  ", len(sales))
print("Distinct customers:  ", sales['Customer_ID'].nunique())
print("Distinct products:   ", sales['Product_ID'].nunique())
print("Distinct stores:     ", sales['Store_ID'].nunique())
```

## 1.4 Initial Structural Observations

- `sales_transactions.csv` is the central fact table (45,000+ rows) joining
  Customers, Products, and Stores.
- Row counts for `Product_ID`, `Store_ID`, and `Customer_ID` referenced inside
  `sales` slightly exceed the counts of unique keys in their respective
  dimension tables — a first signal (confirmed in Notebook 2) that referential
  integrity issues exist and require a formal reconciliation pass before this
  data can be trusted for executive reporting.
- No column-level type coercion issues were observed at load time; all
  numeric fields loaded as `float64`/`int64` as expected.
